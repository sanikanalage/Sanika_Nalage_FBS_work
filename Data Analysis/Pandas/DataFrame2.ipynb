### Concat

```python
df1 = pd.DataFrame({'id':[101,102,103], 'name':['Sanika','Sumit','Arya']})
df1
```

```python
df2 = pd.DataFrame({'dept_id':[1,2,3], 'dname':['HR','Admin','Developer']})
df2
```

```python
pd.concat([df1,df2])
```

#### Merge

```python
emp = pd.DataFrame({'Id':[101,102,103,104], 'Name':['Sanika','Sumit','Arya','Neha'], 'did':[1,2,1,4]})
emp
                   
```

```python
dept=pd.DataFrame({'dept_id':[1,2,3], 'dname':['Admin','IT','Sales']})
dept
```

```python
pd.merge(emp,dept, how='inner')
```

```python
pd.merge(emp,dept, how='left')
```

```python
pd.merge(emp,dept, how='right')
```

```python
pd.merge(emp,dept, how='outer')
```

```python
pd.merge(emp,dept, left_on='did', right_on='dept_id')
```
