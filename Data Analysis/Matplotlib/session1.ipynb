```python
import matplotlib.pyplot as plt
```

#### Bar Plot
1. Comparing categories
2. Showing frequencies or count

```python
products=['Hot Coffe','Cold Coffe','Tea','Salad']
summer = [20,95,80,60]
```

```python
plt.figure(figsize=(7,4))  #(width, height)
plt.bar(products, summer, color='pink')
plt.title('Sales of product for summer')
plt.xlabel('Products')
plt.ylabel('Count')
plt.show()
```
