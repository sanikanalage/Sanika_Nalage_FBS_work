```python
import matplotlib.pyplot as plt
```

#### Lineplot
1. Trends over time
2. Time Series analysis
3. Upward/Downward trend

```python
import numpy as np
```

```python
overs = np.arange(1, 21)
team_1 = np.random.randint(0,37,20)
team_2 = np.random.randint(0,37,20)
```

```python
team_1
```

```python
team1_score = np.cumsum(team_1)
team1_score
```

```python
team2_score = np.cumsum(team_2)
team2_score
```

```python
plt.figure(figsize=(7, 3))
plt.style.use('dark_background')
plt.plot(overs, team1_score, marker='o', color='yellow', label='team1')
plt.plot(overs, team2_score, marker='>', color='blue', label='team2')
plt.title('Match Summary')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.legend()
plt.show()
```

#### Scatterplot
1. Relationship between two variables
2. Correlation
3. Positive Correlation
4. Negative Correlation

```python
plt.figure(figsize=(7,3))
plt.scatter(overs, team_1, marker='o', color='yellow', label='team1')
plt.scatter(overs, team_2, marker='>', color='blue', label='team2')
plt.legend()
plt.show()
```

#### Pie Chart
1. Showing percentage distribution
2. Part to whole comparison
3. Best to use when categories <=5

```python
marks=[98,83,41,47,48,63]
subjects=['Sanskrit','English','Physics','Chemistry','Biology','Maths']
```

```python
plt.pie(marks, labels=subjects, autopct='%0.2f%%')
plt.show()
```
