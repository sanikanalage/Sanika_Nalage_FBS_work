```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text = '''Dr. Smith lives in the U.S. and works at New York University. 
He specializes in A.I. and has published over 3.14 papers every year. 
He often travels to Washington D.C. for conferences, workshops, etc. 
Recently, he received an award from the A.I. Society.'''
```

```python
doc = nlp(text)
```

```python
doc
```

```python
type(doc)
```

```python
for sen in doc.sents:
    print(sen)
```

```python
for word in doc:
    print(word, ':' , word.pos_)
```

### Entity Recognition

```python
txt = 'BCCI is board of Indian Cricket'
```

```python
txt = 'India is a developing country, Mumbai is economical capital. Mr.Fadanvis is Cheif Minister of Maharashtra.'
```

```python
doc = nlp(txt)
```

```python
for token in doc.ents:
    print(token, ':',token.label_, spacy.explain(token.label_))
```

#### Stop words removal

```python
stop_words = nlp.Defaults.stop_words
```

```python
len(stop_words)
```

```python
stop_words
```

```python
import re
pattern = r'[A-Za-z]+'
res = re.findall(pattern, txt)
res
doc = nlp(' '.join(res))
```

```python
result = []
for word in doc:
    word = str(word)
    if word.lower() not in stop_words:
        result.append(word)
```
