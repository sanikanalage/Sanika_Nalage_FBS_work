```python
X = data.drop("Species", axis=1)
y = data['Species']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=12, test_size=0.2)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC()
svc.fit(X_train, y_train)
```

```python
train_pred = svc.predict(X_train)
test_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print('Training accuracy:', accuracy_score(y_train, train_pred))
print('Testing accuracy:', accuracy_score(y_test, test_pred))
```

### To draw the hyperplane using sctterplot & lineplot we need only 2 variable

```python
data.columns
```

```python
data['Species'].unique()
```

```python
data['Species'] = data['Species'].replace({'Setosa':0, 'Versicolor':1, 'Virginica':2}).astype('int')
```

```python
X = data.drop(["Species",'Sepal_Length','Sepal_Width'], axis=1).values
y = data['Species'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=12, test_size=0.2)
```

```python
svc = SVC()
svc.fit(X_train, y_train)
```

```python
train_pred = svc.predict(X_train)
test_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print('Training accuracy:', accuracy_score(y_train, train_pred))
print('Testing accuracy:', accuracy_score(y_test, test_pred))
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X_train, y_train,svc)
```
