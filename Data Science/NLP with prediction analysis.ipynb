```python
txt = 'Name: Review, Length: 1000, dtype: str'
pattern = r'[^A-Za-z]'
res = re.sub(pattern, ' ' ,txt)
res
```

```python
processed_list = []
pattern = r'[^A-Za-z]'
for i in range(corpus.shape[0]):
    row = corpus[i]
    txt = re.sub(pattern, ' ' ,row)
    doc = nlp(txt)
    result = []
    for token in doc:
        if token.lemma_ not in stop_words:
            result.append(token.lemma_)
    result = ' '.join(result)
    processed_list.append(result)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
obj = CountVectorizer()
result = obj.fit_transform(processed_list)
result = result.toarray()
result
```

#### Gaussian Naive Bayes: Classification algorithm based on Bayes' Theorem
1. Naive:Assume the features are independent of each other.
2. Bayes:Uses Bayestheorem to calculate probability
3. Gaussian:Numbers follows normal distribution for each class.

```python
X = result
y = data['Liked']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=7, test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model = GaussianNB()
model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)
pred
```

```python
train_pred = model.predict(X_train)
train_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

```python
accuracy_score(y_train, train_pred)
```
