```python
import pandas as pd
```

```python
data = pd.read_csv("C:/Users/DELL/Desktop/FBS/Data Science/Datasets/Iris.csv")
data.head()
```

```python
data.info()
```

```python
data["Species"].value_counts()
```

```python
X = data.drop("Species",axis=1)
y = data['Species']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test,y_train, y_test = train_test_split(X,y,random_state = 10, test_size = 0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier()
```

```python
dt.fit(X_train,y_train)
```

```python
train_pred = dt.predict(X_train)
test_pred = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print('Trianing accuracy:', accuracy_score(y_train, train_pred))
print('Testing accuracy:', accuracy_score(y_test, test_pred))
```

```python
from sklearn.tree import plot_tree
```

```python
import matplotlib.pyplot as plt
```

```python
plot_tree(dt, class_names = data['Species'].unique(),feature_names=X.columns,filled=True)
plt.show()
```
